General notebook for Pandas + Matplotlib support

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/materials_qc_raw.csv')
class Plotter:
    def __init__(self, data: pd.DataFrame):
        self.data = data
        self.x_measurement = None
        self.y_measurement = None

    def simple_plot(self, x_measurement: str, y_measurement: str):
        self.x_measurement = x_measurement
        self.y_measurement = y_measurement
        self._plot(self.data)
        return self

    def plot_by_cat(self, category: str, cat_value: str):
        if self.x_measurement is None or self.y_measurement is None:
            raise ValueError("Call simple_plot before plot_by_cat.")

        filtered_data = self.data.loc[self.data[category] == cat_value]
        self._plot(filtered_data)

    def _plot(self, data: pd.DataFrame):
        plt.plot(data[self.x_measurement], data[self.y_measurement], 'o')
        plt.show()

plotter = Plotter(df)
plotter.simple_plot("Tensile Strength (MPa)", "Yield Strength (MPa)")
plotter.plot_by_cat("Material ", "Aluminum 6061")
    
```
