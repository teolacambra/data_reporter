General notebook for Pandas + Matplotlib support

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('steel_tensile_data.csv')
class Plotter:
    def __init__(self, data:pd.DataFrame, x_measurement:str, y_measurement:str):
        self.data = data
        self.x_measurement = data[x_measurement]
        self.y_measurement = data[y_measurement]

    def simple_plot(self):
      
        plt.plot(self.x_measurement, self.y_measurement, 'o')
        plt.show()
steel_tensile_vs_elong = Plotter(df, 'elongation', 'tensile').simple_plot()
    
```
